```python
import pandas as pd

# Diccionario base de usuarios y citas
usuarios = {}
citas = {}

def agregar_usuario():
    cedula = input("Ingrese la cédula del paciente: ")
    if cedula in usuarios:
        print("El usuario ya está registrado.")
        return
    nombre = input("Ingrese el nombre del paciente: ")
    edad = input("Ingrese la edad del paciente: ")
    telefono = input("Ingrese el número de teléfono del paciente: ")
    usuarios[cedula] = {"nombre": nombre, "edad": edad, "teléfono": telefono}
    print("Usuario agregado con éxito.")

def agregar_cita():
    cedula = input("Ingrese la cédula del paciente: ")
    if cedula not in usuarios:
        print("El usuario no está registrado. Por favor, agréguelo primero.")
        return
    fecha = input("Ingrese la fecha de la cita (YYYY-MM-DD): ")
    hora = input("Ingrese la hora de la cita (HH:MM): ")
    tipo = input("Ingrese el tipo de cita: ")
    medico = input("Ingrese el nombre del médico: ")
    cita = {"fecha": fecha, "hora": hora, "tipo": tipo, "médico": medico, "estado": "Pendiente"}
    if cedula in citas:
        citas[cedula].append(cita)
    else:
        citas[cedula] = [cita]
    print("Cita agregada con éxito.")

def confirmar_cancelar_cita():
    cedula = input("Ingrese la cédula del paciente: ")
    if cedula not in citas or not citas[cedula]:
        print("No hay citas registradas para este paciente.")
        return
    print("Citas registradas:")
    for i, cita in enumerate(citas[cedula]):
        print(f"{i + 1}. Fecha: {cita['fecha']}, Hora: {cita['hora']}, Médico: {cita['médico']}, Estado: {cita['estado']}")
    seleccion = int(input("Seleccione el número de la cita para modificar: ")) - 1
    if seleccion < 0 or seleccion >= len(citas[cedula]):
        print("Selección inválida.")
        return
    estado = input("Ingrese el nuevo estado (Confirmada/Cancelada): ").capitalize()
    if estado in ["Confirmada", "Cancelada"]:
        citas[cedula][seleccion]["estado"] = estado
        print("Estado de la cita actualizado.")
    else:
        print("Estado inválido.")

def generar_reporte():
    print("Usuarios registrados:")
    for cedula, datos in usuarios.items():
        print(f"{cedula} - {datos['nombre']}, {datos['edad']} años, Tel: {datos['teléfono']}")
    print("\nCitas registradas:")
    for cedula, lista_citas in citas.items():
        for cita in lista_citas:
            print(f"{cedula} - {cita['fecha']} {cita['hora']} {cita['tipo']} con {cita['médico']} - Estado: {cita['estado']}")

def menu():
    while True:
        print("\n--- Menú Principal ---")
        print("1. Agregar usuario")
        print("2. Agregar cita")
        print("3. Confirmar o cancelar cita")
        print("4. Generar reporte")
        print("5. Salir")
        opcion = input("Seleccione una opción: ")
        if opcion == "1":
            agregar_usuario()
        elif opcion == "2":
            agregar_cita()
        elif opcion == "3":
            confirmar_cancelar_cita()
        elif opcion == "4":
            generar_reporte()
        elif opcion == "5":
            print("Saliendo del sistema. ¡Hasta luego!")
            break
        else:
            print("Opción inválida. Intente de nuevo.")

if __name__ == "__main__":
    menu()
```
